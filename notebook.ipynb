```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import drive

# Mount Google Drive
drive.mount('/content/drive')

# --- Task 1: Dataset Understanding ---

# TODO: IMPORTANT! Replace 'None' with the actual LOCAL path to your dataset file.


dataset_path = "https://drive.google.com/drive/folders/16lHceA9Y0e3BMD6Ru3sOl7PP7yO2_s50"

if dataset_path is None:
    print("Error: Dataset path is not set. Please update the 'dataset_path' variable after copying the dataset to your Drive.")
    df = None
else:
    try:
        df = pd.read_csv(dataset_path)
        print(f"Dataset loaded successfully from {dataset_path}!")
    except FileNotFoundError:
        print(f"Error: Dataset not found at {dataset_path}. Please check the path.")
        print("Make sure the dataset from the shared folder is in your Google Drive.")
        df = None # Initialize df as None to prevent further errors if file is not found
    except Exception as e:
        print(f"An error occurred while loading the dataset: {e}")
        df = None

if df is not None:
    print("\n--- Dataset Information ---")
    print(f"Number of records: {len(df)}")
    print("\nSample records:")
    print(df.head())

    print("\nAvailable columns:")
    print(df.columns.tolist())

    # TODO: Update 'text_column' and 'target_column' based on your dataset's column names
    # Common names for text include 'text', 'review', 'sentence', 'tweet'
    # Common names for target include 'sentiment', 'label', 'category'
    text_column = 'text' # CHANGE THIS IF YOUR TEXT COLUMN IS NAMED DIFFERENTLY
    target_column = 'sentiment' # CHANGE THIS IF YOUR TARGET COLUMN IS NAMED DIFFERENTLY

    if text_column in df.columns:
        df['text_length'] = df[text_column].apply(len)
        print(f"\nAverage text length: {df['text_length'].mean():.2f}")
        print(f"Min text length: {df['text_length'].min()}")
        print(f"Max text length: {df['text_length'].max()}")
    else:
        print(f"Warning: Text column '{text_column}' not found. Cannot analyze text length.")

    if target_column in df.columns:
        print("\nTarget labels/classes:")
        print(df[target_column].unique())
        print("\nClass distribution:")
        print(df[target_column].value_counts())

        # Plotting class distribution
        plt.figure(figsize=(8, 6))
        sns.countplot(x=target_column, data=df)
        plt.title('Class Distribution')
        plt.xlabel('Class')
        plt.ylabel('Count')
        plt.show()
    else:
        print(f"Warning: Target column '{target_column}' not found. Cannot analyze class distribution.")
else:
    print("Cannot proceed with dataset analysis as the dataset was not loaded due to the error above.")
```
